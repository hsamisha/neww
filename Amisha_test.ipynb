```python
# ============================================================
# E-COMMERCE SALES DATA ANALYSIS
# END-TO-END EXPLORATORY DATA ANALYSIS
# ============================================================

# ============================================================
# STEP 1 — IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

from IPython.display import display

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")


# ============================================================
# STEP 2 — LOAD DATASET
# ============================================================

# IMPORTANT:
# Change this path if your CSV is located somewhere else.

file_path = "/content/sales_data/sales_dataset/sales_dataset.csv"

# If your CSV is directly inside /content, use:
# file_path = "/content/sales_dataset.csv"

df = pd.read_csv(file_path)

print("\nDataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# ============================================================
# STEP 3 — INITIAL DATA EXPLORATION
# ============================================================

print("\n========== FIRST 5 ROWS ==========")
display(df.head())

print("\n========== LAST 5 ROWS ==========")
display(df.tail())

print("\n========== RANDOM SAMPLE ==========")
display(df.sample(5))

print("\n========== DATASET SHAPE ==========")
print(df.shape)

print("\n========== COLUMN NAMES ==========")
print(df.columns.tolist())

print("\n========== DATA TYPES ==========")
print(df.dtypes)

print("\n========== DATA INFORMATION ==========")
df.info()

print("\n========== STATISTICAL SUMMARY ==========")
display(df.describe(include='all').T)


# ============================================================
# STEP 4 — DATA QUALITY CHECK
# ============================================================

print("\n========== MISSING VALUES ==========")

missing_values = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Values': missing_values,
    'Missing Percentage': missing_percentage.round(2)
})

missing_df = missing_df.sort_values(
    by='Missing Values',
    ascending=False
)

display(missing_df)


print("\n========== DUPLICATES ==========")

duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)


print("\n========== UNIQUE VALUES ==========")

unique_values = df.nunique().sort_values()

display(unique_values)


# ============================================================
# STEP 5 — REMOVE DUPLICATES
# ============================================================

if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicates removed.")
else:
    print("No duplicate rows found.")

print("Current dataset shape:", df.shape)


# ============================================================
# STEP 6 — CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('-', '_')
)

print("\n========== CLEANED COLUMN NAMES ==========")
print(df.columns.tolist())


# ============================================================
# STEP 7 — DATA TYPE CLEANING
# ============================================================

# Date
if 'date' in df.columns:
    df['date'] = pd.to_datetime(
        df['date'],
        errors='coerce'
    )

# Amount
if 'amount' in df.columns:
    df['amount'] = pd.to_numeric(
        df['amount'],
        errors='coerce'
    )

# Quantity
# Your dataset uses 'qty'
if 'qty' in df.columns:
    df['qty'] = pd.to_numeric(
        df['qty'],
        errors='coerce'
    )

print("\nData types after cleaning:")
print(df.dtypes)


# ============================================================
# STEP 8 — HANDLE MISSING AMOUNT VALUES
# ============================================================

if 'amount' in df.columns:

    amount_median = df['amount'].median()

    print("\nMedian Amount:", amount_median)

    df['amount'] = df['amount'].fillna(
        amount_median
    )

    print(
        "Missing Amount values after cleaning:",
        df['amount'].isnull().sum()
    )


# ============================================================
# STEP 9 — HANDLE QUANTITY
# ============================================================

if 'qty' in df.columns:

    print("\n========== QUANTITY ANALYSIS ==========")

    display(df['qty'].describe())

    print(
        "Missing Quantity values:",
        df['qty'].isnull().sum()
    )

    print(
        "Quantity <= 0:",
        (df['qty'] <= 0).sum()
    )


# ============================================================
# STEP 10 — CREATE DATE FEATURES
# ============================================================

if 'date' in df.columns:

    df['year'] = df['date'].dt.year
    df['month'] = df['date'].dt.month
    df['month_name'] = df['date'].dt.month_name()
    df['day'] = df['date'].dt.day
    df['day_name'] = df['date'].dt.day_name()

    df['day_type'] = np.where(
        df['date'].dt.dayofweek >= 5,
        'Weekend',
        'Weekday'
    )

    print("\nDate features created successfully.")


# ============================================================
# STEP 11 — KEY BUSINESS KPIs
# ============================================================

print("\n========================================")
print("          BUSINESS KPIs")
print("========================================")

total_orders = df['order_id'].nunique()
total_sales = df['amount'].sum()
total_quantity = df['qty'].sum()
average_order_value = df['amount'].mean()

print(f"Total Orders        : {total_orders:,}")
print(f"Total Sales         : ₹{total_sales:,.2f}")
print(f"Total Quantity      : {total_quantity:,}")
print(f"Average Order Value : ₹{average_order_value:,.2f}")


# ============================================================
# STEP 12 — MONTHLY SALES TREND
# ============================================================

print("\n========== MONTHLY SALES ==========")

monthly_sales = (
    df.groupby(
        df['date'].dt.to_period('M')
    )['amount']
    .sum()
    .reset_index()
)

monthly_sales['date'] = (
    monthly_sales['date'].dt.to_timestamp()
)

display(monthly_sales)

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_sales,
    x='date',
    y='amount',
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales Amount')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# Highest and lowest sales month

highest_sales_month = monthly_sales.loc[
    monthly_sales['amount'].idxmax()
]

lowest_sales_month = monthly_sales.loc[
    monthly_sales['amount'].idxmin()
]

print(
    "Highest Sales Month:",
    highest_sales_month['date'].strftime('%B %Y')
)

print(
    "Highest Sales Amount: ₹{:,.2f}".format(
        highest_sales_month['amount']
    )
)

print(
    "Lowest Sales Month:",
    lowest_sales_month['date'].strftime('%B %Y')
)

print(
    "Lowest Sales Amount: ₹{:,.2f}".format(
        lowest_sales_month['amount']
    )
)


# ============================================================
# STEP 13 — SALES BY CATEGORY
# ============================================================

print("\n========== SALES BY CATEGORY ==========")

category_sales = (
    df.groupby('category')['amount']
    .sum()
    .sort_values(ascending=False)
)

display(category_sales)

plt.figure(figsize=(12, 6))

category_sales.sort_values().plot(
    kind='barh'
)

plt.title('Sales by Product Category')
plt.xlabel('Sales Amount')
plt.ylabel('Category')

plt.tight_layout()
plt.show()


top_category = category_sales.idxmax()
lowest_category = category_sales.idxmin()

print("Top Category:", top_category)
print(
    "Top Category Sales: ₹{:,.2f}".format(
        category_sales.max()
    )
)

print("Lowest Category:", lowest_category)
print(
    "Lowest Category Sales: ₹{:,.2f}".format(
        category_sales.min()
    )
)

category_percentage = (
    category_sales /
    category_sales.sum() * 100
).round(2)

print("\nCategory Sales Contribution (%):")
display(category_percentage)


# ============================================================
# STEP 14 — TOP 10 PRODUCTS
# ============================================================

print("\n========== TOP 10 PRODUCTS ==========")

top_products = (
    df.groupby('style')['amount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

display(top_products)

plt.figure(figsize=(12, 6))

top_products.sort_values().plot(
    kind='barh'
)

plt.title('Top 10 Products by Sales')
plt.xlabel('Sales Amount')
plt.ylabel('Product Style')

plt.tight_layout()
plt.show()

best_product = top_products.idxmax()

print("Best-Selling Product:", best_product)

print(
    "Best Product Sales: ₹{:,.2f}".format(
        top_products.max()
    )
)

top_10_contribution = (
    top_products.sum() /
    df['amount'].sum() * 100
)

print(
    f"Top 10 products contribute "
    f"{top_10_contribution:.2f}% of total sales."
)


# ============================================================
# STEP 15 — STATE-WISE SALES
# ============================================================

print("\n========== TOP STATES BY SALES ==========")

state_sales = (
    df.groupby('ship_state')['amount']
    .sum()
    .sort_values(ascending=False)
)

display(state_sales.head(10))

plt.figure(figsize=(12, 6))

state_sales.head(10).sort_values().plot(
    kind='barh'
)

plt.title('Top 10 States by Sales')
plt.xlabel('Sales Amount')
plt.ylabel('State')

plt.tight_layout()
plt.show()

top_state = state_sales.idxmax()

print("Top State:", top_state)

print(
    "Top State Sales: ₹{:,.2f}".format(
        state_sales.max()
    )
)


# ============================================================
# STEP 16 — STATE-WISE ORDERS
# ============================================================

state_orders = (
    df.groupby('ship_state')['order_id']
    .nunique()
    .sort_values(ascending=False)
)

print("\nTop 10 States by Orders:")
display(state_orders.head(10))

plt.figure(figsize=(12, 6))

state_orders.head(10).sort_values().plot(
    kind='barh'
)

plt.title('Top 10 States by Number of Orders')
plt.xlabel('Number of Orders')
plt.ylabel('State')

plt.tight_layout()
plt.show()


# ============================================================
# STEP 17 — B2B VS B2C
# ============================================================

print("\n========== B2B VS B2C ==========")

print("Customer type distribution:")
display(df['b2b'].value_counts(dropna=False))

b2b_sales = (
    df.groupby('b2b')['amount']
    .sum()
    .sort_values(ascending=False)
)

b2b_orders = (
    df.groupby('b2b')['order_id']
    .nunique()
)

b2b_quantity = (
    df.groupby('b2b')['qty']
    .sum()
)

b2b_summary = pd.DataFrame({
    'Orders': b2b_orders,
    'Quantity': b2b_quantity,
    'Sales': b2b_sales
})

display(b2b_summary)

plt.figure(figsize=(8, 5))

b2b_sales.plot(kind='bar')

plt.title('B2B vs B2C Sales')
plt.xlabel('Customer Type')
plt.ylabel('Sales Amount')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()

b2b_sales_percentage = (
    b2b_sales /
    b2b_sales.sum() * 100
).round(2)

print("Sales Contribution (%):")
display(b2b_sales_percentage)


# ============================================================
# STEP 18 — SALES CHANNEL
# ============================================================

print("\n========== SALES CHANNEL ==========")

sales_channel = (
    df.groupby('sales_channel')['amount']
    .sum()
    .sort_values(ascending=False)
)

display(sales_channel)

plt.figure(figsize=(8, 5))

sales_channel.plot(kind='bar')

plt.title('Sales by Sales Channel')
plt.xlabel('Sales Channel')
plt.ylabel('Sales Amount')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()

channel_orders = (
    df.groupby('sales_channel')['order_id']
    .nunique()
    .sort_values(ascending=False)
)

channel_summary = pd.DataFrame({
    'Orders': channel_orders,
    'Sales': sales_channel
})

display(channel_summary)


# ============================================================
# STEP 19 — ORDER STATUS
# ============================================================

print("\n========== ORDER STATUS ==========")

status_counts = df['status'].value_counts()

display(status_counts)

plt.figure(figsize=(10, 6))

status_counts.plot(kind='bar')

plt.title('Order Status Distribution')
plt.xlabel('Order Status')
plt.ylabel('Number of Records')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

status_percentage = (
    df['status']
    .value_counts(normalize=True) * 100
).round(2)

print("Status Percentage:")
display(status_percentage)


# ============================================================
# STEP 20 — CANCELLATION ANALYSIS
# ============================================================

print("\n========== CANCELLATION ANALYSIS ==========")

cancelled_df = df[
    df['status']
    .astype(str)
    .str.lower()
    .str.contains('cancel')
]

print(
    "Cancelled Records:",
    len(cancelled_df)
)

cancellation_rate = (
    len(cancelled_df) /
    len(df) * 100
)

print(
    f"Cancellation Rate: "
    f"{cancellation_rate:.2f}%"
)


cancel_category = (
    cancelled_df['category']
    .value_counts()
    .sort_values(ascending=False)
)

print("\nCancellations by Category:")
display(cancel_category)

plt.figure(figsize=(12, 6))

cancel_category.plot(kind='bar')

plt.title('Cancellations by Product Category')
plt.xlabel('Category')
plt.ylabel('Number of Cancelled Records')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


cancel_state = (
    cancelled_df
    .groupby('ship_state')['order_id']
    .nunique()
    .sort_values(ascending=False)
    .head(10)
)

print("\nTop States by Cancelled Orders:")
display(cancel_state)


# ============================================================
# STEP 21 — FULFILLMENT ANALYSIS
# ============================================================

print("\n========== FULFILLMENT ANALYSIS ==========")

fulfillment_counts = (
    df['fulfilment']
    .value_counts()
)

display(fulfillment_counts)

fulfillment_sales = (
    df.groupby('fulfilment')['amount']
    .sum()
    .sort_values(ascending=False)
)

display(fulfillment_sales)

plt.figure(figsize=(8, 5))

fulfillment_sales.plot(kind='bar')

plt.title('Sales by Fulfillment Method')
plt.xlabel('Fulfillment Method')
plt.ylabel('Sales Amount')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# STEP 22 — SHIPPING SERVICE LEVEL
# ============================================================

print("\n========== SHIPPING SERVICE LEVEL ==========")

shipping_counts = (
    df['ship_service_level']
    .value_counts()
)

display(shipping_counts)

shipping_sales = (
    df.groupby('ship_service_level')['amount']
    .sum()
    .sort_values(ascending=False)
)

display(shipping_sales)

plt.figure(figsize=(8, 5))

shipping_sales.plot(kind='bar')

plt.title('Sales by Shipping Service Level')
plt.xlabel('Shipping Service Level')
plt.ylabel('Sales Amount')

plt.tight_layout()
plt.show()


# ============================================================
# STEP 23 — PRODUCT DEMAND
# ============================================================

print("\n========== PRODUCT DEMAND ==========")

category_quantity = (
    df.groupby('category')['qty']
    .sum()
    .sort_values(ascending=False)
)

display(category_quantity)

plt.figure(figsize=(12, 6))

category_quantity.sort_values().plot(
    kind='barh'
)

plt.title('Product Demand by Category')
plt.xlabel('Quantity Sold')
plt.ylabel('Category')

plt.tight_layout()
plt.show()


# ============================================================
# STEP 24 — TOP PRODUCTS BY QUANTITY
# ============================================================

top_quantity_products = (
    df.groupby('style')['qty']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\nTop 10 Products by Quantity:")
display(top_quantity_products)

plt.figure(figsize=(12, 6))

top_quantity_products.sort_values().plot(
    kind='barh'
)

plt.title('Top 10 Products by Quantity Sold')
plt.xlabel('Quantity Sold')
plt.ylabel('Product Style')

plt.tight_layout()
plt.show()


# ============================================================
# STEP 25 — SIZE ANALYSIS
# ============================================================

print("\n========== SIZE ANALYSIS ==========")

size_quantity = (
    df.groupby('size')['qty']
    .sum()
    .sort_values(ascending=False)
)

display(size_quantity)

plt.figure(figsize=(10, 6))

size_quantity.plot(kind='bar')

plt.title('Quantity Sold by Product Size')
plt.xlabel('Size')
plt.ylabel('Quantity Sold')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# STEP 26 — PROMOTION ANALYSIS
# ============================================================

print("\n========== PROMOTION ANALYSIS ==========")

df['has_promotion'] = df['promotion_ids'].notna()

promotion_sales = (
    df.groupby('has_promotion')['amount']
    .agg(['count', 'sum', 'mean'])
)

display(promotion_sales)

plt.figure(figsize=(8, 5))

promotion_sales['sum'].plot(kind='bar')

plt.title('Sales: Promoted vs Non-Promoted Orders')
plt.xlabel('Promotion Applied')
plt.ylabel('Sales Amount')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# STEP 27 — CURRENCY ANALYSIS
# ============================================================

print("\n========== CURRENCY ANALYSIS ==========")

currency_counts = (
    df['currency']
    .value_counts()
)

display(currency_counts)

currency_sales = (
    df.groupby('currency')['amount']
    .sum()
    .sort_values(ascending=False)
)

display(currency_sales)


# ============================================================
# STEP 28 — CITY ANALYSIS
# ============================================================

print("\n========== TOP CITIES ==========")

city_sales = (
    df.groupby('ship_city')['amount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

display(city_sales)

plt.figure(figsize=(12, 6))

city_sales.sort_values().plot(
    kind='barh'
)

plt.title('Top 10 Cities by Sales')
plt.xlabel('Sales Amount')
plt.ylabel('City')

plt.tight_layout()
plt.show()


# ============================================================
# STEP 29 — OUTLIER ANALYSIS
# ============================================================

print("\n========== OUTLIER ANALYSIS ==========")

plt.figure(figsize=(10, 5))

sns.boxplot(x=df['amount'])

plt.title('Amount Distribution and Outliers')
plt.xlabel('Amount')

plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 5))

sns.boxplot(x=df['qty'])

plt.title('Quantity Distribution and Outliers')
plt.xlabel('Quantity')

plt.tight_layout()
plt.show()


# ============================================================
# STEP 30 — DISTRIBUTION ANALYSIS
# ============================================================

plt.figure(figsize=(10, 5))

sns.histplot(
    df['amount'],
    bins=50,
    kde=True
)

plt.title('Distribution of Order Amount')
plt.xlabel('Amount')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 5))

sns.histplot(
    df['qty'],
    bins=30,
    kde=True
)

plt.title('Distribution of Quantity')
plt.xlabel('Quantity')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()


# ============================================================
# STEP 31 — CORRELATION ANALYSIS
# ============================================================

print("\n========== CORRELATION ANALYSIS ==========")

numeric_data = df[
    ['qty', 'amount']
]

correlation = numeric_data.corr()

display(correlation)

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f'
)

plt.title('Correlation Between Numerical Variables')

plt.tight_layout()
plt.show()


# ============================================================
# STEP 32 — MONTHLY QUANTITY TREND
# ============================================================

monthly_quantity = (
    df.groupby(
        df['date'].dt.to_period('M')
    )['qty']
    .sum()
    .reset_index()
)

monthly_quantity['date'] = (
    monthly_quantity['date'].dt.to_timestamp()
)

display(monthly_quantity)

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_quantity,
    x='date',
    y='qty',
    marker='o'
)

plt.title('Monthly Quantity Sold Trend')
plt.xlabel('Month')
plt.ylabel('Quantity Sold')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# STEP 33 — YEARLY SALES
# ============================================================

print("\n========== YEARLY SALES ==========")

yearly_sales = (
    df.groupby('year')['amount']
    .sum()
)

display(yearly_sales)

plt.figure(figsize=(8, 5))

yearly_sales.plot(kind='bar')

plt.title('Yearly Sales')
plt.xlabel('Year')
plt.ylabel('Sales Amount')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# STEP 34 — WEEKDAY VS WEEKEND
# ============================================================

print("\n========== WEEKDAY VS WEEKEND ==========")

day_type_sales = (
    df.groupby('day_type')['amount']
    .sum()
)

display(day_type_sales)

plt.figure(figsize=(8, 5))

day_type_sales.plot(kind='bar')

plt.title('Sales: Weekday vs Weekend')
plt.xlabel('Day Type')
plt.ylabel('Sales Amount')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# STEP 35 — FINAL KPI TABLE
# ============================================================

print("\n========================================")
print("             FINAL KPIs")
print("========================================")

final_kpis = pd.DataFrame({
    'Metric': [
        'Total Orders',
        'Total Sales',
        'Total Quantity Sold',
        'Average Order Value',
        'Cancellation Rate (%)',
        'Number of Categories',
        'Number of Products',
        'Number of States',
        'Number of Cities'
    ],

    'Value': [
        df['order_id'].nunique(),
        df['amount'].sum(),
        df['qty'].sum(),
        df['amount'].mean(),
        cancellation_rate,
        df['category'].nunique(),
        df['style'].nunique(),
        df['ship_state'].nunique(),
        df['ship_city'].nunique()
    ]
})

display(final_kpis)


# ============================================================
# STEP 36 — FINAL DATA QUALITY CHECK
# ============================================================

print("\n========================================")
print("       FINAL DATA QUALITY CHECK")
print("========================================")

print("Final Dataset Shape:", df.shape)

print("\nMissing Values:")
display(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())


# ============================================================
# STEP 37 — AUTOMATIC KEY INSIGHTS
# ============================================================

print("\n========================================")
print("             KEY INSIGHTS")
print("========================================")

print(
    f"1. The dataset contains "
    f"{df.shape[0]:,} records and "
    f"{df.shape[1]:,} columns after cleaning."
)

print(
    f"2. Total sales recorded in the dataset are "
    f"₹{df['amount'].sum():,.2f}."
)

print(
    f"3. The dataset contains "
    f"{df['order_id'].nunique():,} unique orders."
)

print(
    f"4. Total quantity sold is "
    f"{df['qty'].sum():,} units."
)

print(
    f"5. The average order value is "
    f"₹{df['amount'].mean():,.2f}."
)

print(
    f"6. The highest-selling product category is "
    f"'{category_sales.idxmax()}'."
)

print(
    f"7. The highest-selling product/style is "
    f"'{top_products.idxmax()}'."
)

print(
    f"8. The highest-sales state is "
    f"'{state_sales.idxmax()}'."
)

print(
    f"9. The cancellation rate is "
    f"{cancellation_rate:.2f}%."
)

print(
    f"10. The top 10 products contribute "
    f"{top_10_contribution:.2f}% of total sales."
)

print(
    f"11. The highest-sales month is "
    f"{highest_sales_month['date'].strftime('%B %Y')}."
)

print(
    f"12. The lowest-sales month is "
    f"{lowest_sales_month['date'].strftime('%B %Y')}."
)


# ============================================================
# STEP 38 — RECOMMENDATIONS
# ============================================================

print("\n========================================")
print("            RECOMMENDATIONS")
print("========================================")

print("""
1. Inventory Planning
   Use historical sales and quantity trends to maintain adequate
   inventory for high-demand products and categories.

2. Cancellation Monitoring
   Investigate products and categories with relatively high
   cancellation activity to identify potential operational issues.

3. Fulfillment Optimization
   Monitor fulfillment methods and shipping service levels to
   identify opportunities for improving delivery efficiency.

4. Customer Segmentation
   Analyze B2B and B2C customers separately to understand their
   purchasing patterns and develop targeted strategies.

5. Product Management
   Prioritize high-demand products for inventory monitoring and
   replenishment planning.

6. Geographic Strategy
   Use state and city-level sales patterns to support regional
   marketing and distribution decisions.

7. Promotion Evaluation
   Compare promoted and non-promoted orders to understand how
   promotional activity relates to sales performance.

8. KPI Monitoring
   Regularly monitor sales, orders, average order value,
   cancellation rate, product demand, and fulfillment metrics.
""")


# ============================================================
# STEP 39 — CONCLUSION
# ============================================================

print("\n========================================")
print("               CONCLUSION")
print("========================================")

print("""
This project performed an end-to-end exploratory data analysis
of e-commerce sales data.

The analysis covered data cleaning, missing-value treatment,
data quality checks, descriptive statistics, financial
performance, customer behavior, geographical sales patterns,
product demand, fulfillment, shipping, promotions,
cancellations, and order-status analysis.

The findings provide a data-driven view of sales, customers,
products, and operational activity. These insights can support
inventory planning, customer targeting, fulfillment management,
and monitoring of cancellation and return patterns.

Overall, the analysis demonstrates how exploratory data analysis
can transform raw e-commerce data into meaningful business
insights and support data-driven decision-making.
""")


# ============================================================
# STEP 40 — FINAL DATASET PREVIEW
# ============================================================

print("\n========================================")
print("        FINAL CLEANED DATASET")
print("========================================")

display(df.head())

print("\nAnalysis completed successfully!")
```
